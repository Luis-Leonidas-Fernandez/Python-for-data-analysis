### What this example is about

Write the main idea here before running your own version.


````python
b = arr[[1,5,7]] [:, [0,3,1,2]]
# fila 0 → [ 0  1  2  3]
# fila 1 → [ 4  5  6  7]  ←
# fila 2 → [ 8  9 10 11]
# ...
# fila 5 → [20 21 22 23]  ←
# ...
# fila 7 → [28 29 30 31]  ←

# Obtenes temporalmente:
# [[ 4  5  6  7]
# [20 21 22 23]
# [28 29 30 31]]

# : significa todas las filas del array temporal
# Dame las columnas en este orden: 0, luego 3, luego 1, y finalmente 2

print("devuelve las filas 1,5,7 y reordena las columnas de todas esas filas:", b)
````

### What this example is about

Write the main idea here before running your own version.


````python
# ¿Cuántas de las 5.000 caminatas llegaron alguna vez a una 
# distancia de 30 o más respecto del origen?
hits30 = (np.abs(walks) >= 30).any(axis=1)
print("hits :", hits30)
sumhits = hits30.sum()
# abs() elimina el signo:
# nos interesa la distancia respecto de 0, independientemente de la dirección.

# >= 30
# Esto genera otra matriz (5000, 1000), pero de booleanos.
# axis=1 trabaja horizontalmente, es decir, sobre cada fila/caminata.

# Para cada caminata pregunta:

# ¿Existe al menos un True entre sus 1.000 pasos?

# hits30.sum()
# [False, True, True, False, True]

#   0  +  1  + 1  +  0  + 1

#              = 3

# walks
#   ↓
# np.abs(walks)
#   ↓
# distancias al origen
#   ↓
# >= 30
#   ↓
# matriz de True/False (5000 × 1000)
#   ↓
# .any(axis=1)
#   ↓
# un True/False por caminata (5000,)
#   ↓
# .sum()
#   ↓
# 3395 caminatas
print("hit sum:", sumhits)
````

## Example 111

### Original code

```python
crossing_times = (np.abs(walks[hits30]) >= 30).argmax(axis=1)
crossing_times
```

### What this example is about

Write the main idea here before running your own version.


````python
# Para cada una de las 3.395 caminatas que sabemos que llegaron a ±30,
#  ¿en qué índice ocurrió por primera vez?
crossing_times = (np.abs(walks[hits30]) >= 30).argmax(axis=1)
# walks[hits30]
# Antes obtuviste:
# hits30:
# [False, True, True, ..., True, False, True]
# Eso funciona como máscara sobre las filas de walks.
# walks[hits30] significa: Quedate solamente con las caminatas
#  que llegaron a ±30.
# np.abs(walks[hits30]) >= 30 significa:
# para cada una de esas 3.395 caminatas genera algo conceptualmente así:
# caminata A:
# [F F F F F ... F T T F ...]
#                  ↑
#             llegó a ±30

# caminata B:
# [F F F ... F F F F ... T ...]
#                          ↑
#                     llegó a ±30
# Cada fila tiene 1000 booleanos.

# argmax(axis=1) significa:
# Ahora NumPy busca el primer True de cada fila.
# argmax() encuentra el índice del primer máximo, así que encuentra 
# el primer True.

# Y axis=1 significa que lo hace horizontalmente, independientemente 
# para cada caminata:
print("cruce de tiempo:", crossing_times)
````
